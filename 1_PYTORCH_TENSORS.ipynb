Tensors are a specialized data structure that are very similar to arrays and matrices. In PyTorch, we use tensors to encode the inputs and outputs of a model,
as well as the model’s parameters.

Tensors are similar to NumPy’s ndarrays, except that tensors can run on GPUs or other specialized hardware to accelerate computing.
If you’re familiar with ndarrays, you’ll be right at home with the Tensor API. If not, follow along in this quick API walkthrough.

```python
import torch
import numpy as np
```

Tensor Initialization
Tensors can be initialized in various ways. Take a look at the following examples:



**1.Directly from data**  

torch.tensor(data)

```python
data = [[1,2],[3,4]]
x_data = torch.tensor(data)
print(x_data)
```

**2.From a NumPy array**   
 torch.from_numpy(np_array)

```python
np_array = np.array(data)
print(np_array)
x_np = torch.from_numpy(np_array)
print(x_np)
```

### Difference 1: Math Behavior
Watch what happens when we try to multiply our data structures by 2.

```python
# Python list duplicates elements
print("List * 2:", data * 2)

# NumPy array and PyTorch Tensor perform element-wise multiplication
print("NumPy * 2:\n", np_array * 2)
print("Tensor * 2:\n", x_data * 2)
```

### Difference 2: Device (CPU vs GPU)
Tensors can be sent to other processors (like a GPU) to run parallel calculations instantly.

```python
# Check if GPU is available
if torch.cuda.is_available():
    gpu_tensor = x_data.to('cuda')
    print("Tensor moved to GPU:", gpu_tensor)
    print("Device:", gpu_tensor.device)
else:
    print("Currently running on CPU device:", x_data.device)
    print("(You can enable GPUs in Colab via Edit -> Notebook settings)")
```

**3.From another tensor:**

```python
x_ones = torch.ones_like(x_data) # retains the properties of x_data The new tensor retains the properties (shape, datatype) of the argument tensor, unless explicitly overridden.
x_rand = torch.rand_like(x_data, dtype=torch.float) # overrides the datatype of x_data

print(x_ones)
print(x_rand)
```

**4.With random or constant values:PROVIDING JUST SHAPE**

```python
shape = (2, 3, )
rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zero_tensor = torch.zeros(shape)

print(f"Random Tensor: \n {rand_tensor} \n")
print(f"Ones Tensor: \n {ones_tensor} \n")
print(f"Ones Tensor: \n {zero_tensor} \n")
```

***Tensor Attributes***

```python
tensor = torch.rand(3, 4)

print(f"Shape of tensor: {tensor.shape}")
print(f"Datatype of tensor: {tensor.dtype}")
print(f"Device tensor is stored on: {tensor.device}")
```

**Tensor Operations**

```python
# We move our tensor to the GPU if available
```

```python
if torch.accelerator.is_available():
  device = torch.accelerator.current_accelerator().type
else:
  'cpu'
tensor.to(device)
print(f"Device tensor is stored on: {tensor.device}")
```

**Standard numpy-like indexing and slicing:**

```python
tensor = torch.ones(4, 4)
tensor[:,1] = 0
print(tensor)
```

**Joining tensors**

You can use torch.cat to concatenate a sequence of tensors along a given dimension.

dim=0: Concatenates vertically (top-to-bottom). It stacks rows on top of each other. The number of rows increases

dim=1: Concatenates horizontally (left-to-right). It places columns next to each other. The number of columns increases.

```python
t1 = torch.cat([tensor, tensor, tensor], dim=1)
print(t1)
```

```python
t2 = torch.cat([tensor, tensor, tensor], dim=0)
print("Vertically Concatenated (dim=0) Shape:", t2.shape)
print(t2)
```

**Multiplying tensors**

```python
a = tensor * tensor
print (a)
```

```python
# matrix multiplication between two tensors
a = rand_tensor.matmul(zero_tensor.T)
print(a)
```

```python
#In-place operations Operations that have a _ suffix are in-place. For example: x.copy_(y), x.t_(), will change x.
print(tensor, "\n")
tensor.add_(5)
print(tensor)

```

**Tensor to NumPy array**

```python
t = torch.ones(5)
print(f"t: {t}")
n = t.numpy()
print(f"n: {n}")
```
